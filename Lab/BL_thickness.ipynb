```python
import numpy as np
```

BL thickness for oscillatory flow:
$$\delta = \sqrt{\frac{2\nu}{\omega}}$$

```python
# Frequenscies in Hertz
freqs_20 = [1.1, 1.2, 1.3, 1.4, 1.5, 1.6, 1.7, 1.8]
freqs_30 = [0.898, 0.980, 1.061, 1.143, 1.225, 1.306, 1.388, 1.470]
```

```python
def stokes_boundary_layer(f):
    nu = 1e-6  # Dynamic viscosity of water in kg/(m·s)
    d = np.sqrt(2*nu/(2*np.pi*f))
    return d
```

```python
d_20 = [stokes_boundary_layer(f) for f in freqs_20]
d_30 = [stokes_boundary_layer(f) for f in freqs_30]
```

```python
print('Boundary layer thickness for frequencies at 20 cm water depth:')
for d in d_20:
    print(f'{d*1000:.1f} mm')

print('Boundary layer thickness for frequencies at 30 cm water depth:')
for d in d_30:
    print(f'{d*1000:.1f} mm')   
```
