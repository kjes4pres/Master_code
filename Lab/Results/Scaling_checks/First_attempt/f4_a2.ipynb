## Comparing the experiments across depth by nondimensionalization

```python
def make_dimensionless(eta, k, omega, t):
    """
    Convert amplitude to dimensionless form.
    """
    eta_star = eta * k
    t_star = t * omega
    
    return eta_star, t_star
```

```python
eta_non_20_R1_P0, t_non_20 = make_dimensionless(r1_20['P_0'], k_20, 2 * np.pi * freq1, r1_20['time'])
eta_non_30_R1_P0, t_non_30 = make_dimensionless(r1_30['P_0'], k_30, 2 * np.pi * freq2, r1_30['time'])

eta_non_20_R2_P0, _ = make_dimensionless(r2_20['P_0'], k_20, 2 * np.pi * freq1, r2_20['time'])
eta_non_30_R2_P0, _ = make_dimensionless(r2_30['P_0'], k_30, 2 * np.pi * freq2, r2_30['time'])

eta_non_20_R3_P0, _ = make_dimensionless(r3_20['P_0'], k_20, 2 * np.pi * freq1, r3_20['time'])
eta_non_30_R3_P0, _ = make_dimensionless(r3_30['P_0'], k_30, 2 * np.pi * freq2, r3_30['time'])

fig, ax = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

ax[0].plot(eta_non_20_R1_P0, label='20 cm - R1')
ax[0].plot(eta_non_30_R1_P0, label='30 cm - R1')

ax[1].plot(eta_non_20_R2_P0, label='20 cm - R2')
ax[1].plot(eta_non_30_R2_P0, label='30 cm - R2')

ax[2].plot(eta_non_20_R3_P0, label='20 cm - R3')
ax[2].plot(eta_non_30_R3_P0, label='30 cm - R3')

ax[0].legend()
ax[1].legend()
ax[2].legend()

ax[2].set_xlabel('T*')
ax[0].set_ylabel('η*')
ax[1].set_ylabel('η*')
ax[2].set_ylabel('η*')

fig.suptitle('Dimensionless Surface Elevation at Probe 0 for Both Depths', fontsize=16)
```
